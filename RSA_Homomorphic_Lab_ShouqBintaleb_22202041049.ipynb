```python
import math

# function to find modular inverse (private key d)
def find_private_key(e, phi):
    for d in range(1, phi):
        if (e * d) % phi == 1:
            return d
    return None
# STEP 1: RSA KEY GENERATION
# Choose two prime numbers
p = 11
q = 13
print(f"p = {p}, q = {q}")

# Calculate n and Euler's totient function
n = p * q
phi = (p-1) * (q-1)
print(f"n = {n}, φ = {phi}")

# public/private keys
e = 7
print(f"Public key (e, n) = ({e}, {n})")
d = find_private_key(e, phi)
print(f"Private key (d, n) = ({d}, {n})")


# STEP 2: ENCRYPTION AND DECRYPTION FUNCTIONS
def encrypt(message, e, n):
    """
    Encrypt a message using RSA: ciphertext = message^e mod n
    """
    return pow(message, e, n)

def decrypt(ciphertext, d, n):
    """
    Decrypt a ciphertext using RSA: message = ciphertext^d mod n
    """
    return pow(ciphertext, d, n)

# STEP 3: ENCRYPT TWO MESSAGES
# plaintext messages
m1 = 5
m2 = 9
print(f"\nm1 = {m1}, m2 = {m2}")

# Encrypt both messages
c1 = encrypt(m1, e, n)
c2 = encrypt(m2, e, n)
print(f"E(m1) = {c1}, E(m2) = {c2}")

# STEP 4: DEMONSTRATE HOMOMORPHIC PROPERTY
# Multiply the two ciphertexts
cipher_multiply = (c1 * c2) % n
print(f"E(m1)*E(m2) mod n = {cipher_multiply}")

# Decrypt the multiplied ciphertext
decrypted_result = decrypt(cipher_multiply, d, n)
print(f"Decrypted result = {decrypted_result}")
expected_result = (m1 * m2) % n
print(f"Expected (m1*m2 mod n) = {expected_result}")

# STEP 5: FINAL VERIFICATION
print(f"\nHomomorphic property verified: E(m1)E(m2) ≡ E(m1*m2) mod n")

```
